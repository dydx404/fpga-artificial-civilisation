# DMA Test

Starter notebook for validating packed frame transfers. The current code runs in dry-run loopback mode.

```python
from pathlib import Path
import sys
import numpy as np

sys.path.append(str(Path('..').resolve() / 'python'))
from dma_interface import DmaFrameInterface, pack_strategy_frame, unpack_strategy_frame

height, width = 8, 8
strategies = np.random.default_rng(1).integers(0, 2, size=(height, width), dtype=np.uint8)
packed = pack_strategy_frame(strategies)
dma = DmaFrameInterface(dry_run=True)
returned = dma.transfer_frame(packed)
unpacked = unpack_strategy_frame(returned, height, width)
assert np.array_equal(strategies, unpacked)
unpacked
```
